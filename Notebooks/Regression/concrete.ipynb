```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
current_dir = Path.cwd()
file_path = current_dir.parent.parent / 'Datasets' / 'Regression datasets' / 'Concrete Data' / 'Concrete_Data.xls'

df = pd.read_excel(file_path, engine='xlrd')

df.head(10)
```

```python
new_columns =  [
    "cement",
    "slag",
    "fly_ash",
    "water",
    "superplasticizer",
    "coarse_aggregate",
    "fine_aggregate",
    "age",
    "strength"
]


df.columns = new_columns
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
X = df.drop('strength', axis=1)
y = df['strength']
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
import pickle
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=10, test_size=0.25)
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
scaler_path = Path.cwd().parent.parent/ "ML_models and scalers" / "Scalers" / "concrete_scaler.pkl"

pickle.dump(scaler, open(scaler_path, 'wb'))
```

```python
## LinearRegression

model = LinearRegression()
model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)

print(f'r2_score : {r2_score(y_test, y_pred)}')
print(f'mean_absolute_error : {mean_absolute_error(y_test, y_pred)}')
print(f'mean_squared_error : {mean_squared_error(y_test, y_pred)}')
```

```python
# Using GridSearchCV

param_grid = {
    'fit_intercept': [True, False],
    'positive': [True, False]
}

grid = GridSearchCV(LinearRegression(), param_grid=param_grid, cv=5, verbose=2)

grid.fit(X_train_scaled, y_train)
grid.best_params_
```

```python
model = LinearRegression(fit_intercept=True, positive=False)
model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)

print(f'r2_score : {r2_score(y_test, y_pred)}')
print(f'mean_absolute_error : {mean_absolute_error(y_test, y_pred)}')
print(f'mean_squared_error : {mean_squared_error(y_test, y_pred)}')
```

```python
model_path = Path.cwd().parent.parent/ "ML_models and scalers" / "ML_Models"/ "Concrete"/"linear_model.pkl"

pickle.dump(model, open(model_path, 'wb'))
```

```python
## Decision Tree Regressor

model = DecisionTreeRegressor()
model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)

print(f'r2_score : {r2_score(y_test, y_pred)}')
print(f'mean_absolute_error : {mean_absolute_error(y_test, y_pred)}')
print(f'mean_squared_error : {mean_squared_error(y_test, y_pred)}')
```

```python
model_path = Path.cwd().parent.parent/ "ML_models and scalers" / "ML_Models"/ "Concrete"/"tree_model.pkl"

pickle.dump(model, open(model_path, 'wb'))
```

```python
# using GridSearchCV

param_grid = {
    'criterion': ['squared_error', 'absolute_error', 'poisson'],
    'splitter': ['best', 'random'],
    'max_depth': [7,8,9,10, 11, 12, 13],
    'min_samples_split': [10, 20, 25, 30, 35],
}

grid = GridSearchCV(DecisionTreeRegressor(),param_grid=param_grid, cv=3, verbose=5, n_jobs=-1)
grid.fit(X_train_scaled, y_train)
print(grid.best_params_)
print(grid.best_score_)
```

```python
model = DecisionTreeRegressor(criterion='squared_error', max_depth=9, min_samples_split=10, splitter='random')
model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)

print(f'r2_score : {r2_score(y_test, y_pred)}')
print(f'mean_absolute_error : {mean_absolute_error(y_test, y_pred)}')
print(f'mean_squared_error : {mean_squared_error(y_test, y_pred)}')
```

```python
## KNeighborsRegressor

model = KNeighborsRegressor()
model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)

print(f'r2_score : {r2_score(y_test, y_pred)}')
print(f'mean_absolute_error : {mean_absolute_error(y_test, y_pred)}')
print(f'mean_squared_error : {mean_squared_error(y_test, y_pred)}')
```

```python
param_grid = {
    'n_neighbors':[5,7,9,11,13],
    'algorithm': ['auto', 'ball_tree', 'kd_tree', 'brute'],
    'weights': ['uniform', 'distance'],
    'metric': ['euclidean', 'manhattan', 'minkowski'],
    'p': [1, 2] # 1 = Manhattan, 2 = Euclidean (only matters if metric='minkowski')
}


grid = GridSearchCV(KNeighborsRegressor(), param_grid=param_grid, cv=5, verbose=5)
grid.fit(X_train_scaled, y_train)
grid.best_params_
```

```python
model = KNeighborsRegressor(algorithm='brute', metric='euclidean', n_neighbors=5, p=1, weights='distance')
model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)

print(f'r2_score : {r2_score(y_test, y_pred)}')
print(f'mean_absolute_error : {mean_absolute_error(y_test, y_pred)}')
print(f'mean_squared_error : {mean_squared_error(y_test, y_pred)}')
```

```python
model_path = Path.cwd().parent.parent/ "ML_models and scalers" / "ML_Models"/ "Concrete"/"kNN_model.pkl"

pickle.dump(model, open(model_path, 'wb'))
```

```python
## Support Vector Regressor (SVR)

model = SVR()
model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)

print(f'r2_score : {r2_score(y_test, y_pred)}')
print(f'mean_absolute_error : {mean_absolute_error(y_test, y_pred)}')
print(f'mean_squared_error : {mean_squared_error(y_test, y_pred)}')
```

```python
param_grid = {
    "kernel": ["linear", "rbf", "poly"],
    "C": [0.1, 1, 10, 100],
    "epsilon": [0.01, 0.1, 0.2, 0.5],
    "gamma": ["scale", "auto"]
}

grid = GridSearchCV(SVR(), param_grid=param_grid, cv=5, verbose=5)
grid.fit(X_train_scaled, y_train)
grid.best_params_
```

```python
model = SVR(C=100, epsilon=0.5, gamma='scale', kernel='rbf')
model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)

print(f'r2_score : {r2_score(y_test, y_pred)}')
print(f'mean_absolute_error : {mean_absolute_error(y_test, y_pred)}')
print(f'mean_squared_error : {mean_squared_error(y_test, y_pred)}')
```

```python
model_path = Path.cwd().parent.parent/ "ML_models and scalers" / "ML_Models"/ "Concrete"/"svr_model.pkl"

pickle.dump(model, open(model_path, 'wb'))
```

```python
## Ridge Regression

model = Ridge()
model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)

print(f'r2_score : {r2_score(y_test, y_pred)}')
print(f'mean_absolute_error : {mean_absolute_error(y_test, y_pred)}')
print(f'mean_squared_error : {mean_squared_error(y_test, y_pred)}')
```

```python
param_grid = {
    'solver' :['auto', 'svd', 'cholesky', 'lsqr', 'sparse_cg', 'sag', 'saga'],
    'alpha' : np.logspace(-4, 4, 50),
    'fit_intercept': [True, False]
}

grid = GridSearchCV(Ridge(), param_grid=param_grid, cv=5, verbose=5)
grid.fit(X_train_scaled, y_train)
grid.best_params_
```

```python
model = Ridge(alpha=0.8286427728546842, fit_intercept=True, solver='svd')
model.fit(X_train_scaled, y_train)
y_pred = model.predict(X_test_scaled)

print(f'r2_score : {r2_score(y_test, y_pred)}')
print(f'mean_absolute_error : {mean_absolute_error(y_test, y_pred)}')
print(f'mean_squared_error : {mean_squared_error(y_test, y_pred)}')
```

```python
model_path = Path.cwd().parent.parent/ "ML_models and scalers" / "ML_Models"/ "Concrete"/"ridge_model.pkl"

pickle.dump(model, open(model_path, 'wb'))
```
